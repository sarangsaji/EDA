BOX PLOT


A box plot also known as box and whisker plot is a graphical representation of the distribution af the data set it disp;ays the data minimum first quartile(Q1),median(Q2),Third quartile(Q3) and maximum values.the box in the box plot shows the interquartile rangeb (IQR).which represents the middle 50% of the data while the whiskers extend to the minimum and maximum values excluding any outliers

COMPONENTS OF A BOX PLOT

* Box represents the middle 50% of the data from (q1 to q3)
* whhiskers extend to the smallest and largest values within 1.5 times the IQR from q1 and q3,respectively
* outliers:points outside the whiskers are considered potential outliers.
* median (Q2) the line inside the box represents the median(50th percentile) of the data.

OUTLIER

An outlier is a data point that differ significially from the rest of the dataset outliers can occur due to varriability in the data or errors in measurment they can skew the results of statistical analyses.so its important to identify  and potentially remove them to  get a clearer understanding of the data
* outliers are the data points that fall outside of the whiskers,typically any value more than 1.5 times the IQR below the first quartile or above the third quartile is considered an outlier

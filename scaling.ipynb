SCALING

In exploratary data analysis scaling refers to process of adjusting the range of features values so they fit within specific range or distribution this is often done to ensure that all features are treated equally by machine or equally by machine learning models

Common types of scalings

min max scaling: this scale the features too fixed range usually[0,1]
the formula for min max scaling is
x scaled=x-min
xmax-xmin
it maintains the relationship between values but forces them within a specific range
